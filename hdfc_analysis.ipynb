```python
mod_data.drop(["symbol", "exchange"], axis=1, inplace=True)
mod_data["timestamp"] = pd.to_datetime(mod_data["timestamp"]).dt.tz_localize(None)
```

```python
def highest_and_lowest_price_in_a_period (mod_data, start_date, end_date) :
    start_date = pd.to_datetime(start_date)
    end_date = pd.to_datetime(end_date)
    req_data = mod_data[["timestamp", "high", "low"]][(mod_data["timestamp"] >= start_date) & (mod_data["timestamp"] < end_date + pd.Timedelta(days=1))].copy()
    price_low_high = (req_data["low"].min(), req_data["high"].max())
    return price_low_high

def average_price_in_a_period (mod_data, start_date, end_date) :
    start_date = pd.to_datetime(start_date)
    end_date = pd.to_datetime(end_date)
    req_data = mod_data[["timestamp", "high", "low"]][(mod_data["timestamp"] >= start_date) & (mod_data["timestamp"] < end_date + pd.Timedelta(days=1))].copy()
    req_data["average"] = (req_data["high"] + req_data["low"]) / 2
    avg_price = req_data["average"].mean()
    return avg_price

def total_returns_in_a_period (mod_data, start_date, end_date, inv_amt) :
    start_date = pd.to_datetime(start_date)
    end_date = pd.to_datetime(end_date)
    req_data = mod_data[["timestamp", "open", "close"]][(mod_data["timestamp"] >= start_date) & (mod_data["timestamp"] < end_date + pd.Timedelta(days=1))].copy()
    start = req_data["open"].iloc[0]
    end = req_data["close"].iloc[-1]
    total_return_percentage = ((end - start) / start) * 100
    total_profit_amount = inv_amt * total_return_percentage / 100
    final_data = (total_return_percentage, total_profit_amount)
    return final_data

def avg_volume_calculator (mod_data) :
    avg_vol = mod_data["volume"].mean()
    return avg_vol

def volatility_in_a_period (mod_data, start_date, end_date) :
    start_date = pd.to_datetime(start_date)
    end_date = pd.to_datetime(end_date)
    req_data = mod_data[["timestamp", "close"]][(mod_data["timestamp"] >= start_date) & (mod_data["timestamp"] < end_date + pd.Timedelta(days=1))].copy()
    req_data = req_data.reset_index(drop=True)
    req_returns = []
    no_rows = req_data.shape[0]
    for index, items in req_data["close"].items() :
        if (index != (no_rows - 1)) :
            min_wise_return = (req_data.loc[(index+1), "close"] - req_data.loc[index, "close"]) / req_data.loc[index, "close"] * 100
            req_returns.append(min_wise_return)
        else :
            continue
    volatility = np.std(req_returns)
    return volatility
```

```python
print("For the maximum and minimum price in a period, enter 1")
print("For the average price in a period, enter 2")
print("For total returns in a period, enter 3")
print("For average volume, enter 4")
print("For volatility in a period, enter 5")

choice = int (input ("Enter your choice : "))

match choice :
    case 1 :
        start_date = input ("Enter the start date in specified format (YYYY-MM-DD) : ")
        end_date = input ("Enter the end date in specified format (YYYY-MM-DD) : ")
        result = highest_and_lowest_price_in_a_period (mod_data, start_date, end_date)
        print(f"The lowest price in the given period : {result[0]}")
        print(f"The highest price in the given period : {result[1]}")
    case 2 :
        start_date = input ("Enter the start date in specified format (YYYY-MM-DD) : ")
        end_date = input ("Enter the end date in specified format (YYYY-MM-DD) : ")
        result = average_price_in_a_period (mod_data, start_date, end_date)
        print(f"The average price in the period is : {result:.2f}")
    case 3 :
        start_date = input ("Enter the start date in specified format (YYYY-MM-DD) : ")
        end_date = input ("Enter the end date in specified format (YYYY-MM-DD) : ")
        inv_amt = int (input ("Enter the initial invested amount : "))
        result = total_returns_in_a_period (mod_data, start_date, end_date, inv_amt)
        print(f"The total return percentage in the period is : {result[0]:.2f}%")
        print(f"The total profit amount in the period is : {result[1]:.2f}")
    case 4 :
        result = avg_volume_calculator (mod_data)
        print(f"The average volume of the stock is : {result:.0f}")
    case 5 :
        start_date = input ("Enter the start date in specified format (YYYY-MM-DD) : ")
        end_date = input ("Enter the end date in specified format (YYYY-MM-DD) : ")
        result = volatility_in_a_period (mod_data, start_date, end_date)
        print(f"The volatility in the period is {result:.4f}% per minute")
```
